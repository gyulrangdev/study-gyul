## 0️⃣ 스팸 메일 이야기 — 왜 머신러닝이 필요할까

> 📗 **원문** — "The Problem"

스팸 메일을 걸러내는 프로그램을 만든다고 해 봅시다.

**옛날 방식**은 이렇습니다. 자리에 앉아서 규칙을 하나하나 적는 거예요.

```
규칙 1: '무료 현금'이라는 말이 있으면 → 스팸
규칙 2: 느낌표가 3개보다 많으면 → 스팸
규칙 3: ...
```

몇 주 동안 규칙을 씁니다. 그런데 **스팸 보내는 사람들이 말투를 바꿉니다.**
규칙이 안 통하게 되죠. 그래서 규칙을 더 씁니다. 또 바꿉니다. **끝이 없습니다.**

**머신러닝은 이걸 뒤집습니다.**

규칙을 쓰는 대신, 컴퓨터에게 **답이 적힌 메일 수천 통**을 보여 줍니다.
("이건 스팸", "이건 정상") 그러면 컴퓨터가 **규칙을 스스로 찾아냅니다.**

사람이 생각하지 못했던 패턴까지 찾아내고, 스팸 수법이 바뀌면
**코드를 고치는 대신 새 데이터로 다시 학습시키면** 됩니다.

> ✨ **추가** — 아래 코드로 "규칙 쓰기가 왜 끝이 없는지"를 직접 느껴 봅시다.
> 원문에는 없는 실험입니다.

````python
# 스팸 메일 20개를 흉내 낸 아주 작은 예시입니다.
# (스팸이면 1, 정상이면 0)

메일들 = [
    ("무료 현금 지금 받으세요!!!",        1),
    ("공짜 상품권 클릭!!!",             1),
    ("당첨되셨습니다 지금 확인!!!",       1),
    ("무료 이벤트 참여하세요!!",          1),
    ("내일 회의 자료 첨부합니다",         0),
    ("점심 같이 드실래요?",              0),
    ("보고서 검토 부탁드립니다",          0),
    ("주말 잘 보내세요",                 0),
]

# --- 옛날 방식: 사람이 규칙을 손으로 씁니다 ---
def 규칙으로_판단(문장):
    if "무료" in 문장:
        return 1
    if 문장.count("!") >= 3:
        return 1
    return 0

맞춘개수 = sum(규칙으로_판단(문장) == 정답 for 문장, 정답 in 메일들)
print(f"내가 만든 규칙의 성적: {맞춘개수}/{len(메일들)}개 맞음\n")

for 문장, 정답 in 메일들:
    내판단 = 규칙으로_판단(문장)
    표시 = "⭕" if 내판단 == 정답 else "❌ 틀림"
    print(f"  {표시}  '{문장}'")
````

````python
# 그런데 스팸 보내는 사람이 '무료'를 '무 료'로 바꿔 씁니다.
# 규칙 하나 피하려고 띄어쓰기만 바꾼 거예요.

새로운_스팸 = [
    ("무 료 현금 지금 받으세요",     1),
    ("공 짜 상품권 클릭",           1),
    ("이벤트 당첨 축하드립니다",      1),
]

맞춘개수 = sum(규칙으로_판단(문장) == 정답 for 문장, 정답 in 새로운_스팸)
print(f"새로운 스팸에 대한 성적: {맞춘개수}/{len(새로운_스팸)}개 맞음\n")
for 문장, 정답 in 새로운_스팸:
    표시 = "⭕" if 규칙으로_판단(문장) == 정답 else "❌ 못 잡음"
    print(f"  {표시}  '{문장}'")

print("\n=> 띄어쓰기 하나 바꿨을 뿐인데 규칙이 전부 뚫렸습니다.")
print("   규칙을 또 추가해야 하고, 상대는 또 바꿀 겁니다. 이게 '끝이 없다'는 뜻입니다.")
print("\n   머신러닝은 이 술래잡기를 멈추는 방법입니다.")
````

## 1️⃣ 규칙을 쓰는 대신, 배우게 하기

> 🔗 **앞 절과 이어서** — 0절에서 규칙을 계속 늘려 봤지만 **끝이 없었습니다.**
> 규칙을 하나 막으면 스팸 메일이 하나 더 우회하죠. 이 싸움은 이길 수 없습니다.
> 그래서 접근을 통째로 바꿉니다.

> 📗 **원문** — "Learning From Data, Not Rules"

옛날 프로그래밍과 머신러닝은 **화살표 방향이 반대**입니다.

### 옛날 프로그래밍

```
    규칙  +  데이터   →   결과
   (사람이 씀)
```

사람이 규칙을 씁니다. 프로그램이 그 규칙을 데이터에 적용해서 결과를 냅니다.

### 머신러닝

```
   데이터  +  정답   →   규칙(=모델)
                        (컴퓨터가 찾음)
```

사람은 **데이터와 정답**만 줍니다. 컴퓨터가 **규칙을 찾아냅니다.**

### 🔑 '모델'이라는 게 뭔가요

학습이 끝나면 나오는 결과물을 **모델**이라고 부릅니다.

> **모델 = 컴퓨터가 찾아낸 규칙을 '숫자'로 적어 놓은 것**

사람이 읽을 수 있는 문장이 아니라 **숫자 덩어리**예요.
그리고 이 모델은 **한 번도 본 적 없는 새 데이터**에도 답을 내놓을 수 있습니다.

> ✨ **추가** — 아래 표는 원문의 두 그림을 나란히 비교한 것입니다.

| | 옛날 프로그래밍 | 머신러닝 |
| --- | --- | --- |
| 사람이 주는 것 | **규칙** + 데이터 | 데이터 + **정답** |
| 컴퓨터가 내놓는 것 | 결과 | **규칙 (= 모델)** |
| 상황이 바뀌면 | **코드를 고친다** | **다시 학습시킨다** |
| 비유 | 요리법을 알려 주기 | 요리 사진 수천 장 보여 주기 |

## 2️⃣ 머신러닝의 세 가지 종류

> 🔗 **앞 절과 이어서** — "배우게 하자"까지는 좋습니다. 그런데 **뭘 주고 배우게 하죠?**
> 무엇을 쥐여 주느냐에 따라 머신러닝이 세 갈래로 나뉩니다.

> 📗 **원문** — "The Three Types of Machine Learning"

### ① 지도 학습 (Supervised Learning)

**답이 적힌 문제집으로 공부하는 것**과 같습니다.

문제(사진)와 답(고양이/강아지)이 **함께** 주어집니다.
컴퓨터가 문제를 풀고 → 답과 맞춰 보고 → 틀리면 고칩니다. 수천 번 반복하면 요령이 생깁니다.

- "고양이/강아지 라벨이 붙은 사진 1만 장이야. 구별하는 법을 배워."
- "집의 특징과 가격이야. 가격을 맞히는 법을 배워."

### ② 비지도 학습 (Unsupervised Learning)

**답이 없는 문제집**입니다. 입력만 있고 정답이 없어요.

컴퓨터가 **스스로 규칙성을 찾습니다.**

- "손님 1만 명의 구매 기록이야. 비슷한 사람들끼리 묶어 봐." (정답이 뭔지는 아무도 모릅니다)

### ③ 강화 학습 (Reinforcement Learning)

**게임하면서 배우는 것**입니다.

행동을 하면 **점수(보상)** 나 **벌점**을 받습니다. 점수를 최대로 얻는 방법을 스스로 찾습니다.

- "게임을 해 봐. 이기면 +1점, 지면 −1점. 이기는 방법을 찾아내."

### 🔑 실무에서는

- **지도 학습**이 압도적으로 많습니다 (여러분이 만들 것의 대부분)
- **비지도 학습**은 데이터를 살펴보거나 준비할 때 씁니다
- **강화 학습**은 게임 AI, 로봇, 그리고 **ChatGPT의 RLHF**에 쓰입니다

> ✨ **추가** — 세 종류를 한 장의 표로 정리했습니다. 원문에는 없습니다.

| | 지도 학습 | 비지도 학습 | 강화 학습 |
| --- | --- | --- | --- |
| **주어지는 것** | 문제 + **정답** | 문제만 | 환경 + **점수** |
| **비유** | 답지 있는 문제집 | 답 없이 분류하기 | 게임하며 배우기 |
| **묻는 것** | "정답이 뭐야?" | "뭐가 비슷해?" | "어떻게 해야 이겨?" |
| **예시** | 스팸 분류, 집값 예측 | 고객 묶기 | 바둑 AI, 로봇 |

````python
# ✨ 추가 — 어떤 종류인지 맞혀 보는 연습입니다. 원문에는 없습니다.

문제들 = [
    ("사진을 보고 고양이인지 강아지인지 맞히기",       "지도"),
    ("손님들을 비슷한 취향끼리 그룹으로 묶기",         "비지도"),
    ("로봇이 넘어지지 않고 걷는 법을 익히기",          "강화"),
    ("어제 날씨로 내일 기온 맞히기",                  "지도"),
    ("뉴스 기사 10만 개에서 비슷한 주제 찾아내기",     "비지도"),
    ("게임에서 최고 점수 내는 방법 찾기",              "강화"),
]

print("어떤 종류일까요? (정답은 바로 아래에)\n")
힌트 = {"지도": "정답이 함께 주어짐",
        "비지도": "정답 없이 규칙성만 찾음",
        "강화": "점수를 받으며 시도해 봄"}

for 문제, 정답 in 문제들:
    print(f"  {문제}")
    print(f"     -> {정답} 학습   ({힌트[정답]})\n")

print("💡 구별하는 요령:")
print("   '정답이 있나?' -> 있으면 지도, 없으면 비지도")
print("   '해 보면서 점수를 받나?' -> 그러면 강화")
````

## 3️⃣ 분류와 회귀

> 🔗 **앞 절과 이어서** — 2절의 세 가지 중 **지도 학습**을 더 파고듭니다.
> 이 단원부터 5단원까지가 전부 지도 학습이거든요.
> 지도 학습 안에서도 **맞혀야 할 답이 '종류'냐 '숫자'냐**에 따라 갈립니다.

> 📗 **원문** — "Classification vs Regression"

지도 학습은 다시 **두 가지**로 나뉩니다. 이 구분이 아주 중요합니다.

### 분류 (Classification) — "어느 쪽이야?"

답이 **정해진 몇 개 중 하나**입니다.

- 이 메일은 스팸인가 아닌가? → 2개 중 하나
- 이 사진은 고양이/강아지/새 중 뭔가? → 3개 중 하나

### 회귀 (Regression) — "얼마야?"

답이 **숫자**입니다. 몇이든 될 수 있어요.

- 이 집은 얼마일까? → 3억 2천만원, 4억 1천5백만원, ...
- 내일 기온은? → 23.4도

### 한 줄 요약

> **분류는 "어느 칸에 넣을까", 회귀는 "숫자가 몇일까"입니다.**

| | 분류 | 회귀 |
| --- | --- | --- |
| 답의 모양 | 정해진 종류 중 하나 | 아무 숫자나 |
| 예시 | "스팸인가?" | "집값이 얼마인가?" |
| 채점 방법(손실 함수) | 크로스 엔트로피 | 평균제곱오차(MSE) |

> 💡 같은 문제를 **둘 다로 만들 수 있을 때도** 있습니다.
> "주가가 오를까 내릴까?" → 분류
> "주가가 정확히 얼마가 될까?" → 회귀

````python
# ✨ 추가 — 분류와 회귀의 '답 모양'이 어떻게 다른지 눈으로 봅시다.

print("분류 문제의 답 (정해진 것 중 하나)")
분류_예측 = ["스팸", "정상", "정상", "스팸", "정상"]
print(f"  {분류_예측}")
print(f"  가능한 답: {sorted(set(분류_예측))}  <- 딱 2가지뿐\n")

print("회귀 문제의 답 (아무 숫자나)")
회귀_예측 = [32150.5, 41800.0, 28340.7, 55020.3, 37610.9]
print(f"  {회귀_예측}  (단위: 만원)")
print(f"  가능한 답: 무한히 많음  <- 소수점까지 아무 값이나\n")

print("=> 답의 '모양'이 다르니, 채점 방법도 달라야 합니다.")
print("   분류: '맞았나 틀렸나'를 셉니다")
print("   회귀: '얼마나 빗나갔나'를 잽니다")
````

## 4️⃣ 머신러닝 작업 순서

> 🔗 **앞 절과 이어서** — 종류를 다 봤습니다. 이제 **실제로 일할 때의 순서**를 그려 봅니다.
> 이 순서 안에 5절과 6절이 왜 중요한지가 숨어 있습니다.

> 📗 **원문** — "The ML Workflow"

어떤 알고리즘을 쓰든 **작업 순서는 항상 똑같습니다.**

```
  ① 데이터 모으기
        ↓
  ② 청소하고 살펴보기        ← 전체 시간의 60~80%가 여기 (!)
        ↓
  ③ 특징 만들기
        ↓
  ④ 데이터 나누기            ← 5절에서 자세히
        ↓
  ⑤ 학습시키기
        ↓
  ⑥ 평가하기  ──(별로면)──> ③으로 되돌아감
        ↓ (괜찮으면)
  ⑦ 실제 서비스에 넣기
        ↓
  ⑧ 계속 지켜보기 ──(성능 떨어지면)──> ①로 되돌아감
```

### 🔑 놀라운 사실

**②번(데이터 청소)이 전체 시간의 60~80%를 차지합니다.**

영화에서 보는 "멋진 알고리즘 고르기"는 사실 아주 작은 부분이에요.
실제 일의 대부분은 **빠진 값 채우기, 중복 지우기, 이상한 값 찾기**입니다.

### 🔑 또 하나

**③번(특징 만들기)이 알고리즘 선택보다 중요합니다.**

원문 표현을 그대로 옮기면 이렇습니다.

> "좋은 특징이 화려한 알고리즘보다 더 중요하다."

예를 들어 날짜 데이터를 그냥 넣는 것보다 **"무슨 요일인지"** 로 바꿔서 넣으면
훨씬 잘 맞습니다. 주말과 평일은 완전히 다르니까요.

## 5️⃣ 🔑 데이터를 셋으로 나누기 (가장 중요)

> 🔗 **앞 절과 이어서** — 4절 작업 순서의 **두 번째 단계**입니다.
> 그런데 왜 이걸 '가장 중요'라고 붙였을까요? **답은 6절에 있습니다.**
> 지금은 '이렇게 나눈다'만 익히고, 6절에서 '왜 그래야 하는지'를 봅니다.

> 📗 **원문** — "Training, Validation, and Test Splits"
>
> 원문이 이렇게 말합니다: **"초보자가 가장 많이 틀리는 개념"**

### 왜 나눠야 하나요

> **시험 문제를 미리 보고 푼 시험 점수는 실력이 아닙니다.**

컴퓨터도 똑같습니다. **공부한 데이터로 시험을 보면** 점수가 높게 나와요.
그런데 그건 **외운 것**이지 **배운 것**이 아닙니다.

그래서 데이터를 **셋으로 나눕니다.**

| 이름 | 하는 일 | 비유 | 보통 비율 |
| --- | --- | --- | --- |
| **훈련 (Training)** | 컴퓨터가 여기서 배웁니다 | 교과서·문제집 | 60~80% |
| **검증 (Validation)** | 설정을 고를 때 씁니다 | 모의고사 | 10~20% |
| **테스트 (Test)** | 마지막에 **딱 한 번** 실력 확인 | 수능 | 10~20% |

### ⚠️ 테스트 데이터는 '성역'입니다

원문 표현: **"The test set is sacred. You look at it exactly once."**
(테스트 세트는 신성하다. 딱 한 번만 본다.)

테스트 점수를 보고 모델을 계속 고치면, 그건 **테스트 데이터로 공부한 것**과 같습니다.
그러면 그 점수는 **아무 의미가 없어집니다.**

> 💡 데이터가 아주 적으면 **k-겹 교차검증**을 씁니다.
> 데이터를 k조각으로 나눠서, k−1조각으로 배우고 1조각으로 시험 보기를
> 돌아가며 반복한 뒤 평균을 냅니다.

> ✨ **추가** — "공부한 문제로 시험 보면 어떻게 되는지"를 직접 실험해 봅니다.
> 원문에는 설명만 있고 이 실험은 없습니다.

````python
# ✨ 추가 실험 — 외우기만 하는 '가짜 학습기'를 만들어 봅니다.
# 이 학습기는 아무것도 배우지 않고, 본 문제와 답을 통째로 외우기만 합니다.

class 외우기만_하는_학습기:
    def 공부하기(self, 문제들, 답들):
        # 문제와 답을 그냥 사전에 저장 (= 통째로 외우기)
        self.외운것 = {tuple(문제): 답 for 문제, 답 in zip(문제들, 답들)}

    def 답하기(self, 문제들):
        결과 = []
        for 문제 in 문제들:
            if tuple(문제) in self.외운것:
                결과.append(self.외운것[tuple(문제)])    # 외운 문제면 정답
            else:
                결과.append(0)                          # 처음 보면 그냥 0이라고 찍음
        return np.array(결과)

# 데이터 만들기 (2차원 점 200개, 정답은 0 또는 1)
rng = np.random.default_rng(0)
문제 = rng.normal(size=(200, 2))
답 = (문제[:, 0] + 문제[:, 1] > 0).astype(int)

# 훈련용 140개 / 테스트용 60개로 나누기
훈련문제, 훈련답 = 문제[:140], 답[:140]
테스트문제, 테스트답 = 문제[140:], 답[140:]

학습기 = 외우기만_하는_학습기()
학습기.공부하기(훈련문제, 훈련답)

훈련점수 = (학습기.답하기(훈련문제) == 훈련답).mean()
테스트점수 = (학습기.답하기(테스트문제) == 테스트답).mean()

print("'외우기만 하는 학습기'의 성적표")
print(f"  공부한 문제로 시험 (훈련):  {훈련점수:.1%}   <- 만점!")
print(f"  처음 보는 문제로 시험 (테스트): {테스트점수:.1%}   <- 찍은 것과 비슷")
print("\n=> 훈련 점수만 보면 '완벽한 AI'처럼 보입니다.")
print("   하지만 새 문제에서는 아무것도 못 합니다. 외웠을 뿐이니까요.")
print("\n   ⭐ 이래서 반드시 '처음 보는 데이터'로 시험을 봐야 합니다.")
````

## 6️⃣ 과적합과 과소적합

> 🔗 **앞 절과 이어서** — **5절이 왜 필요했는지, 여기가 답입니다.**
> 데이터를 안 나누면 **모델이 답을 외워 버려도 알아챌 방법이 없습니다.**
> 5절과 6절은 사실 한 덩어리예요.

> 📗 **원문** — "Overfitting vs Underfitting"

방금 본 '외우기 학습기'의 문제에는 이름이 있습니다. **과적합**이라고 합니다.

### 과소적합 (Underfitting) — 너무 단순해서 못 배움

모델이 **너무 단순**해서 데이터의 패턴을 못 따라갑니다.
휘어 있는 관계를 직선 하나로 맞추려는 것과 같아요.

- 훈련 점수: **나쁨** ❌
- 테스트 점수: **나쁨** ❌

### 과적합 (Overfitting) — 너무 복잡해서 외워 버림

모델이 **너무 복잡**해서 데이터를 통째로 외웁니다. **잡음까지** 외워요.
모든 점을 정확히 지나가는 구불구불한 선을 그리는데, 새 데이터에서는 엉망입니다.

- 훈련 점수: **아주 좋음** ⭕
- 테스트 점수: **나쁨** ❌  ← 이 조합이 과적합의 신호

### 딱 좋음 (Good Fit)

진짜 패턴은 잡고, 잡음은 무시합니다.

- 훈련 점수: 괜찮음 ⭕
- 테스트 점수: 괜찮음 ⭕

### 🔑 진단법

> **훈련 점수는 높은데 테스트 점수가 낮다 → 과적합입니다.**

### 고치는 법

| 문제 | 해결책 |
| --- | --- |
| **과적합** | 데이터 더 모으기 · 모델 단순하게 · 정규화 · 드롭아웃 · 일찍 멈추기 |
| **과소적합** | 모델 복잡하게 · 특징 더 넣기 · 정규화 줄이기 · 더 오래 학습 |

> ✨ **추가 실험** — 세 가지 상태를 눈으로 직접 봅니다. 원문에는 그림 설명만 있습니다.

````python
# ✨ 추가 — 곡선을 몇 차 다항식으로 맞출지에 따라 세 가지가 다 나옵니다.

rng = np.random.default_rng(3)
x_train = np.sort(rng.uniform(-3, 3, 15))
진짜함수 = lambda x: 0.5 * x**2 - 0.3 * x
y_train = 진짜함수(x_train) + rng.normal(0, 0.6, len(x_train))   # 잡음 섞기

x_test = np.sort(rng.uniform(-3, 3, 15))
y_test = 진짜함수(x_test) + rng.normal(0, 0.6, len(x_test))

# 먼저 모든 차수를 계산해 두고, '테스트 오차가 가장 작은 곳'을 기준으로 판정합니다
결과 = []
for 차수 in [1, 2, 3, 5, 9, 14]:
    계수 = np.polyfit(x_train, y_train, 차수)
    훈련오차 = np.mean((np.polyval(계수, x_train) - y_train) ** 2)
    테스트오차 = np.mean((np.polyval(계수, x_test) - y_test) ** 2)
    결과.append((차수, 훈련오차, 테스트오차))

가장좋은차수 = min(결과, key=lambda r: r[2])[0]

print("다항식의 '차수'를 올릴수록 모델이 복잡해집니다.\n")
print(f"{'차수':>6}{'복잡도':>10}{'훈련 오차':>14}{'테스트 오차':>16}{'상태':>14}")
print("-" * 62)
for 차수, 훈련오차, 테스트오차 in 결과:
    if 차수 < 가장좋은차수:
        상태 = "과소적합"
    elif 차수 == 가장좋은차수:
        상태 = "딱 좋음 ⭐"
    else:
        상태 = "과적합"
    복잡도 = "단순" if 차수 <= 2 else ("보통" if 차수 <= 5 else "복잡")
    # 테스트 오차가 엄청나게 커질 수 있어서 지수 표기로 씁니다
    테스트표시 = f"{테스트오차:.3f}" if 테스트오차 < 1000 else f"{테스트오차:.1e}"
    print(f"{차수:>6}{복잡도:>10}{훈련오차:>14.4f}{테스트표시:>16}{상태:>14}")

print(f"\n=> 차수를 올리면 훈련 오차(3번째 칸)는 계속 줄어듭니다. 외우니까요.")
print(f"   그런데 테스트 오차(4번째 칸)는 차수 {가장좋은차수}에서 최소가 되고 그 뒤로 폭발합니다.")
print(f"   차수 14에서는 테스트 오차가 훈련 오차의 {결과[-1][2]/결과[-1][1]:.0e}배입니다.")
print("   이게 '외웠지만 아무것도 배우지 못한' 상태입니다.")
````

## 7️⃣ 🔨 가장 단순한 머신러닝 직접 만들기

> 🔗 **여기서 방향이 바뀝니다** — 6절까지는 전부 개념이었습니다.
> 이제 손을 움직입니다. 4절의 작업 순서를 따라, 5절처럼 데이터를 나누고,
> 6절을 확인하면서 **직접 만들어 봅니다.**

> 📗 **원문** — "Build It: Nearest Centroid Classifier"

이제 진짜 머신러닝을 만들어 봅니다. **세상에서 가장 단순한 방법**으로요.

### 아이디어 — 이름은 어렵지만 내용은 쉽습니다

**최근접 중심 분류기(Nearest Centroid Classifier)** 라고 부릅니다. 하는 일은 이게 전부예요.

```
배울 때:  각 그룹의 '한가운데 점'을 구한다      (그냥 평균입니다)
답할 때:  새 점이 어느 한가운데에 더 가까운지 본다
```

### 🧸 비유

교실에 남학생들과 여학생들이 앉아 있습니다.

1. **남학생들의 평균 자리**를 구합니다 (한가운데)
2. **여학생들의 평균 자리**도 구합니다
3. 새 학생이 들어오면 → **어느 한가운데에 더 가까운지** 봅니다

끝입니다. 이게 머신러닝이에요.

> 💡 **주목할 점:** 반복도, 학습률도, 복잡한 계산도 없습니다.
> 그냥 **평균 두 번**과 **거리 재기**뿐입니다.

> ✨ **추가** — 먼저 아주 작은 숫자로 '손으로' 따라가 봅시다. 원문에는 없습니다.

````python
# ✨ 추가 — 점 6개만 가지고 손으로 계산해 봅니다.

A그룹 = np.array([[1.0, 1.0], [2.0, 1.0], [1.5, 2.0]])   # 오른쪽 위에 모여 있음
B그룹 = np.array([[-1.0, -1.0], [-2.0, -1.5], [-1.5, -2.0]])  # 왼쪽 아래에 모여 있음

print("1단계 — 각 그룹의 '한가운데'를 구합니다 (그냥 평균)\n")
A중심 = A그룹.mean(axis=0)
B중심 = B그룹.mean(axis=0)
print(f"  A그룹 점들: {A그룹.tolist()}")
print(f"     x끼리 평균: ({1.0}+{2.0}+{1.5})/3 = {A중심[0]:.3f}")
print(f"     y끼리 평균: ({1.0}+{1.0}+{2.0})/3 = {A중심[1]:.3f}")
print(f"     -> A의 한가운데 = {A중심}\n")
print(f"  B그룹 점들: {B그룹.tolist()}")
print(f"     -> B의 한가운데 = {B중심}\n")

print("2단계 — 새 점이 어디에 가까운지 봅니다\n")
새점 = np.array([0.5, 0.8])
A까지거리 = np.sqrt(((새점 - A중심) ** 2).sum())
B까지거리 = np.sqrt(((새점 - B중심) ** 2).sum())
print(f"  새로 들어온 점: {새점}")
print(f"     A중심까지 거리: {A까지거리:.3f}")
print(f"     B중심까지 거리: {B까지거리:.3f}")
print(f"     -> 더 가까운 쪽은 {'A' if A까지거리 < B까지거리 else 'B'} 그룹!")
print("\n  (거리 재는 법: 가로차이²+세로차이² 를 더하고 루트 씌우기 = 피타고라스)")
````

````python
# 📗 원문의 NearestCentroid 클래스를 한글 주석과 함께 그대로 구현했습니다.

class 최근접중심_분류기:
    def 학습(self, 문제, 답):
        # 각 그룹의 한가운데(평균)를 구해서 저장합니다
        self.그룹들 = np.unique(답)
        self.중심들 = np.array([문제[답 == g].mean(axis=0) for g in self.그룹들])
        return self

    def 예측(self, 문제):
        # 각 중심까지의 거리를 재서, 가장 가까운 그룹을 답합니다
        거리들 = np.array([
            np.sqrt(((문제 - 중심) ** 2).sum(axis=1))
            for 중심 in self.중심들
        ])
        return self.그룹들[거리들.argmin(axis=0)]

# 📗 원문의 데이터 생성 방식 (숫자만 그대로)
rng = np.random.default_rng(42)
그룹0 = rng.normal(size=(100, 2)) + np.array([1.0, 1.0])
그룹1 = rng.normal(size=(100, 2)) + np.array([-1.0, -1.0])
X = np.vstack([그룹0, 그룹1])
y = np.array([0] * 100 + [1] * 100)

# 훈련용 140개 / 테스트용 60개로 섞어서 나누기
섞기 = rng.permutation(len(X))
X, y = X[섞기], y[섞기]
X훈련, y훈련 = X[:140], y[:140]
X테스트, y테스트 = X[140:], y[140:]

모델 = 최근접중심_분류기().학습(X훈련, y훈련)

print(f"학습이 끝났습니다. 모델이 기억하는 것은 이게 전부입니다:\n")
for g, 중심 in zip(모델.그룹들, 모델.중심들):
    print(f"  그룹 {g}의 한가운데: {중심}")

훈련정답률 = (모델.예측(X훈련) == y훈련).mean()
테스트정답률 = (모델.예측(X테스트) == y테스트).mean()
print(f"\n성적표")
print(f"  훈련 데이터  : {훈련정답률:.1%}")
print(f"  테스트 데이터: {테스트정답률:.1%}   <- 처음 보는 데이터인데도 잘 맞힙니다!")
print("\n=> 아까 '외우기 학습기'와 비교해 보세요.")
print("   그건 테스트에서 무너졌는데, 이건 안 무너집니다.")
print("   외운 게 아니라 '규칙(=중심 위치)'을 배웠기 때문입니다.")
````

## 8️⃣ 기준선과 비교하기

> 🔗 **앞 절과 이어서** — 7절에서 만든 모델이 정확도를 냈습니다. 그런데 **그게 좋은 건가요?**
> "70%"라는 숫자만으로는 아무것도 모릅니다. **비교 대상이 있어야** 알 수 있습니다.

> 📗 **원문** — "Step 3: Compare Against a Baseline"

### 왜 비교해야 하나요

**"정답률 90%!"** 라는 말만으로는 좋은지 나쁜지 알 수 없습니다.

- 동전 던지기로 찍어도 50%가 나오는 문제라면 → 90%는 훌륭합니다
- 그냥 "전부 정상"이라고 답해도 95%가 나오는 문제라면 → 90%는 **형편없습니다**

그래서 항상 **아무 생각 없이 답하는 기준(baseline)** 과 비교해야 합니다.

원문 표현: **"머신러닝 모델이 무작위 추측을 못 이기면, 뭔가 잘못된 것이다."**

````python
# 📗 원문의 baseline 비교 + ✨ 기준선 종류를 더 추가했습니다.

# 기준선 ① 무작위로 찍기
#    한 번만 찍으면 운에 크게 좌우되므로 1000번 반복해 평균을 냅니다
찍기점수들 = [(np.random.default_rng(s).choice([0, 1], size=len(y테스트)) == y테스트).mean()
              for s in range(1000)]
찍기점수 = float(np.mean(찍기점수들))

# ✨ 기준선 ② 무조건 많은 쪽으로 답하기
많은쪽 = np.bincount(y훈련).argmax()
많은쪽점수 = (y테스트 == 많은쪽).mean()

print("성적 비교\n")
print(f"{'방법':<26}{'정답률':>10}{'설명':>30}")
print("-" * 66)
print(f"{'① 무작위로 찍기':<24}{찍기점수:>11.1%}{'동전 던지기 (1000번 평균)':>28}")
print(f"{'② 무조건 많은 쪽':<24}{많은쪽점수:>11.1%}{'가장 흔한 답만 말하기':>30}")
print(f"{'③ 최근접 중심 분류기':<22}{테스트정답률:>11.1%}{'우리가 만든 것':>32}")

최고기준선 = max(찍기점수, 많은쪽점수)
print(f"\n=> 우리 모델이 가장 좋은 기준선({최고기준선:.1%})보다 "
      f"{테스트정답률 - 최고기준선:.1%}p 앞섭니다.")
print("   이 정도면 '진짜로 뭔가 배웠다'고 말할 수 있습니다.")
print(f"\n   ⚠️ ①번을 한 번만 돌리면 60%가 나오기도 합니다. 운이 좋아서요.")
print("      그래서 기준선도 여러 번 재서 평균을 봐야 합니다.")

print("\n⚠️ 만약 ②번이 95%인데 우리 모델이 90%라면?")
print("   숫자만 보면 높아 보여도, 사실 아무 쓸모가 없는 모델입니다.")
print("   그래서 정답률만 보면 안 되고 항상 기준선과 비교해야 합니다.")
````

> 그런데 **모든 데이터에 대해 그럴까요?** 아닙니다. 무너지는 모양이 있습니다.

그래서 이런 경우에 실패합니다.

### 🔑 그런데 이게 좋은 소식입니다

| --- | --- | --- |

| 크기 차이 | 특징 스케일링 | 8단원 |

**각 단원이 앞 단원의 한계 위에 세워집니다.**

````python
# ✨ 추가 — 실패하는 경우를 직접 만들어 봅니다. 원문에는 설명만 있습니다.

# 도넛 모양: 안쪽 그룹을 바깥쪽 그룹이 감싸고 있음
rng2 = np.random.default_rng(1)
각도 = rng2.uniform(0, 2*np.pi, 150)
안쪽 = np.c_[rng2.uniform(0, 1.2, 150) * np.cos(각도),
             rng2.uniform(0, 1.2, 150) * np.sin(각도)]
각도2 = rng2.uniform(0, 2*np.pi, 150)
반지름 = rng2.uniform(2.5, 3.5, 150)
바깥 = np.c_[반지름 * np.cos(각도2), 반지름 * np.sin(각도2)]

X도넛 = np.vstack([안쪽, 바깥])
y도넛 = np.array([0]*150 + [1]*150)

도넛모델 = 최근접중심_분류기().학습(X도넛, y도넛)
도넛점수 = (도넛모델.예측(X도넛) == y도넛).mean()

print("도넛 모양 데이터에 우리 분류기를 써 보면?\n")
print(f"  그룹 0(안쪽)의 한가운데: {도넛모델.중심들[0]}")
print(f"  그룹 1(바깥)의 한가운데: {도넛모델.중심들[1]}")
print(f"\n  => 두 중심이 거의 같은 자리에 있습니다! (둘 다 원점 근처)")
print(f"  정답률: {도넛점수:.1%}  <- 찍는 것과 비슷합니다\n")
print("  왜 이럴까요? 바깥 그룹이 원을 빙 둘러 있어서,")
print("  평균을 내면 한가운데(원점)가 나와 버립니다.")
print("  안쪽 그룹의 중심과 겹치니 구별을 못 하는 거죠.")
````

### 준지도 학습 — 답이 일부만 있을 때

정답이 붙은 데이터 100개 + 정답 없는 데이터 10만 개가 있다면?

- **가짜 라벨(pseudo-labeling)**: 100개로 대충 학습 → 나머지에 답을 붙임 → 그걸로 다시 학습
- 컴퓨터가 **스스로 문제집을 만들어 가며** 공부하는 셈입니다

### 자기지도 학습 — 답을 스스로 만들기

**사람이 답을 안 붙여도 됩니다.** 데이터 자체에서 문제를 만들어 냅니다.

- **BERT**: 문장에서 단어 15%를 가리고 → 가린 단어 맞히기 (답은 원래 문장에 있으니까요)
- **GPT**: 앞 단어들을 보고 → 다음 단어 맞히기

> 🔑 **여러분이 쓰는 ChatGPT가 바로 이 방식으로 배웠습니다.**
> 인터넷의 모든 글이 "다음 단어 맞히기" 문제집이 된 거예요.

### 편향-분산 트레이드오프 — 과적합의 수학 버전

6절에서 본 과적합/과소적합을 수학으로 설명한 것입니다.

- **편향(bias)**: 모델이 너무 단순해서 생기는 오차 → **과소적합**
- **분산(variance)**: 데이터가 조금만 바뀌어도 결과가 확 달라지는 것 → **과적합**

둘을 동시에 줄일 수는 없어서, **적당한 지점**을 찾아야 합니다.
(10단원에서 자세히 다룹니다)

### 공짜 점심은 없다 (No Free Lunch)

> **모든 문제에 최고인 알고리즘은 존재하지 않습니다.**

어떤 문제에 잘 맞는 알고리즘은 다른 문제에는 잘 안 맞습니다.
그래서 데이터 과학자들이 **여러 알고리즘을 다 해 보고 비교하는** 겁니다.
